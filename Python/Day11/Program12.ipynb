## Split the data

```python
# create x
# set of independent variables
# is is mandatory to get the x as 2D array or data frame
# axis: 0-> row, 1-> column
# df.drop() will remove the column and return a new dataframe object
# it does not update the existing data frame
x=df.drop('Salary',axis=1)

# create y
# dependent variable
# it is mandatory to get the y as 1D array or series
y=df['Salary']
```

```python
df.head()
```

```python
x.head()
```

```python
from sklearn.model_selection import train_test_split

# split the data into train and test sets
x_train,x_test,y_train,y_test = train_test_split(x,y,train_size=0.8,shuffle=True,random_state=123456)
```

## Model Building

```python
from sklearn.linear_model import LinearRegression

# create the model
model = LinearRegression()

# train the model using x and y
model.fit(x,y)
```

```python
predictions =model.predict([[16]])
print(f"Salary of employee with 16 yrs of exp ={predictions[0]:.2f}")
```

```python
# predict salary of emplyee having 15,17 and 100 years of exp
```

```python
# predict the profit values for test set
y_pred = model.predict(x_test)
```

## Evaluation Metrics

```python
from sklearn.metrics import mean_absolute_error
# mean absolute error (MAE)
mae=mean_absolute_error(y_test,y_pred)
mae
```

```python
from sklearn.metrics import mean_squared_error
#  mean squared error (MSE)
mse=mean_squared_error(y_test,y_pred)
mse
```

```python
from sklearn.metrics import root_mean_squared_error
#  root mean squared error
rmse=root_mean_squared_error(y_test,y_pred)
rmse
```

```python
from sklearn.metrics import r2_score
# get r2 score
r2 =r2_score(y_test,y_pred)
r2
```
