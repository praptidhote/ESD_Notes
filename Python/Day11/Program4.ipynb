## Data Visualization

### Bar chat

```python
# import the required packages

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# load the data
fruits=['apple','blueberry','cherry','oranges']
counts=[40,30,100,55]
bar_colors= ['tab:blue','tab:green','tab:red','tab:orange']
```

```python
plt.bar(fruits,counts,label=fruits,color=bar_colors)

plt.title("Fruit Supply information")
plt.xlabel("Fruits")
plt.ylabel("Quantity")
plt.grid(alpha=0.2)
plt.legend()
```
