```python
# import the required packages

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
import warnings
warnings.filterwarnings(action="ignore")
```

```python
# load the data

labels = 'Frogs','Hogs','Dogs','Logs'
size=[15,30,45,10]
explode=[0,0,0.1,0]
```

```python
plt.pie(size, labels=labels, autopct='%1.1f%%', explode=explode, shadow=True)
plt.legend()
```
