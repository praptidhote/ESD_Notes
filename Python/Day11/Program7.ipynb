### read csv and find correlation between columns using df.corr()

```python
# import the required packages

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# get the data
df=pd.read_csv("Salary_Data.csv")
df.head()
```

```python
df.info()
```

```python
# correlation, to see whether columns are related

df.corr()
```

```python
sns.pairplot(df)
plt.savefig('MyGraph.jpg')
```
