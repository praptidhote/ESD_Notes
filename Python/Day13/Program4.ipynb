```python
x=df.drop(['test_result'],axis=1)

y=df['test_result']
```

```python
from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test = train_test_split(x,y,train_size=0.7,random_state=123456)
```

## Train the Model

### Logistic Model

```python
from sklearn.linear_model import LogisticRegressionCV

def create_model_logistic():
    model=LogisticRegressionCV()
    model.fit(x_train,y_train) 
    return model
```

### KNN

```python
from sklearn.neighbors import KNeighborsClassifier

def create_model_knn():
    model=KNeighborsClassifier()
    model.fit(x_train,y_train) 
    return model
```

### SVM

```python
from sklearn.svm import SVC

def create_model_svm():
    model=SVC()
    model.fit(x_train,y_train) 
    return model
```

### Decision Tree

```python
from sklearn.tree import DecisionTreeClassifier

def create_model_decision():
    model=DecisionTreeClassifier()
    model.fit(x_train,y_train) 
    return model
```

## Random Forest

```python
from sklearn.ensemble import RandomForestClassifier

def create_model_random():
    model=RandomForestClassifier()
    model.fit(x_train,y_train) 
    return model
```

## CAT Boost

```python
from catboost import CatBoostClassifier

def create_model_catboost():
    model=CatBoostClassifier()
    model.fit(x_train,y_train) 
    return model
```

```python
from sklearn.ensemble import AdaBoostClassifier

def create_model_adaboost():
    model=AdaBoostClassifier()
    model.fit(x_train,y_train) 
    return model
```

## Gradient Boost

```python
from sklearn.ensemble import GradientBoostingClassifier

def create_model_gradient():
    model=GradientBoostingClassifier()
    model.fit(x_train,y_train) 
    return model
```

## XGBoost

```python
from xgboost import XGBClassifier

def create_model_xgboost():
    model=XGBClassifier()
    model.fit(x_train,y_train) 
    return model
```

## Model Evaluation

```python
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score

def evaluate_model_on_test(model):
    y_pred = model.predict(x_test)

    accuracy=accuracy_score(y_test,y_pred)
    precision=precision_score(y_test,y_pred)
    recall=recall_score(y_test,y_pred)
    f1=f1_score(y_test,y_pred)
    return f"{accuracy:.2f}",f"{precision:.2f}",f"{recall:.2f}",f"{f1:.2f}"
```
