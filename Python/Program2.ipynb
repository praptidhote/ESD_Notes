## Data Visualization

### Scatter plot

```python
# import the required packages

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# prerequisities
import warnings
warnings.filterwarnings(action="ignore")
```

```python
df = pd.read_csv("/home/sunbeam/Desktop/C_DATA/IoT/iot/Day_11/Code/Salary_Data.csv")
df.head()
```

```python
plt.scatter(df["YearsExperience"],df["Salary"],label="Salary")

#  add the chart with information to make it more readable
plt.title("YearsExperience VS Salary")
plt.xlabel("YearsExperience")
plt.ylabel("Salary")

# show the legend
plt.legend()

plt.grid(axis="y",alpha=0.4)
```
