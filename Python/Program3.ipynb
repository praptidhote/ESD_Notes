## Data Visualization

### Line Plot

```python
# import the required packages

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# load the datset

df = pd.read_csv("/home/sunbeam/Desktop/C_DATA/IoT/iot/Day_11/Code/Salary_Data.csv")
df.head()
```

```python
# Plot the line chart for Salary
plt.plot(df["YearsExperience"], df["Salary"], label="Salary Line", color="red")

# Overlay the scatter points for Salary
plt.scatter(df["YearsExperience"], df["Salary"], label="Salary Points", color="green")

# Add chart information and legend
plt.title("YearsExperience VS Salary")
plt.xlabel("YearsExperience")
plt.ylabel("Salary")
plt.legend()  # This displays the labels on the chart
```
