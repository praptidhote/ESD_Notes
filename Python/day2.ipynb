```python
import numpy as np
```

```python
def function1():
    l1 = [10,20,30,40,50]
    print(l1)
    print(f"Type = {type(l1)}")
    print(f"l1[0] = {type(l1[0])} ")

function1()    
```

```python
def function2():
    n_arr =np.array([11,22,33,44,55])
    print(n_arr)
    print(f"Type = {type(n_arr)}")
    
function2()
```

```python
def function3():
    n_arr = np.array([10,20,30,40,50,60])
    print(n_arr)
    print(f"Type = {type(n_arr)}")
    print(f"DType = {n_arr.dtype}")
    print(f"Item size = {n_arr.itemsize}")
    print(f"Size = {n_arr.size}")
    print(f"nbytes = {n_arr.nbytes}")
    print(f"dim = {n_arr.ndim}")
    
function3()
```

```python
def printProperties(arr):
    print(arr)
    print(f"Type = {type(arr)}")
    print(f"DType = {arr.dtype}")
    print(f"Item size = {arr.itemsize}")
    print(f"Size = {arr.size}")
    print(f"NDIM = {arr.ndim}")
    print(f"NBYTES = {arr.nbytes}")
    
n1 = np.array([2,5,8,9,3,4])
printProperties(n1)
```

```python
n3 = np.array([11,22,33,44],dtype = np.int32)
printProperties(n3)
```

```python
n4= np.array([1.15,2.31,3.8])
printProperties(n4)
```

```python
n5 = np.array(["C","C++","Java","Python"])
printProperties(n5)
```

```python
country=np.array(["India","USA","Australia"])
printProperties(country)
```

```python
arr1=np.arange(10,25,2)
print(arr1)
```

```python
arr1=np.arange(1,10,2)
print(arr1)
```

```python
arr2=np.zeros(5)
print(arr2)
```

```python
arr2=np.ones(5)
print(arr2)
```

```python
n1=np.array([11,22,33,44])
n2=np.array([10,20,30,40])
print(f"N1 = {n1}")
print(f"N2 = {n2}")
print(n1+n2)
```

```python
n1=np.random.random(5)      # generate any random 5 floating-point numbers 
print(n1) 
```

```python
n4=np.random.randint(1,10,4)       # any four integer numbers are randomly generated between 1 to 10       
print(n4) 
```

```python
 # broadcast operation

n1=np.array([11,22,33,44]) 

print(n1*5)
print(n1 + 10) 
print(n1 -10)
```

```python
# slicing-Boolean indexing
n1=np.array([10,20,30,40,50])  
print(f"n1[[false,false,true,true,false]]={n1[[False,False,True,True,False]]}")
```

```python
# filtering
salaries = np.array([10000, 15000, 12000, 30000, 45000, 50000, 55000, 
120000, 150000]) 
print(salaries[salaries > 100000])
```
